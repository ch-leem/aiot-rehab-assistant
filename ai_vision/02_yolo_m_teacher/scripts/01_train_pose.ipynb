```python
import os

os.environ["CUDA_VISIBLE_DEVICES"] = "8"
```

```python
from ultralytics import YOLO
```

```python
model = YOLO("../weights/yolo11m-pose.pt")
```

```python
model.train(
    task="pose",
    data="../dataset/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    name="yolo11m_pose_21points",
    project="runs/pose",
    workers=8,
    patience=20,
    optimizer="AdamW",
    lr0=0.001,
    cos_lr=True,
)
```
