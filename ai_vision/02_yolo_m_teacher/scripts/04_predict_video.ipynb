```python
import os

os.environ["CUDA_VISIBLE_DEVICES"] = "8"
```

```python
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("../runs/pose/yolo11m_pose_21points/weights/best.pt")
```

```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("웹캠 열기 실패")

print("웹캠 시작 (중단: q)")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, conf=0.3, device=0, verbose=False)
    annotated = results[0].plot()

    cv2.imshow("YOLO11m Pose Webcam", annotated)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
model.predict(
    source=0,          # webcam
    stream=True,
    conf=0.25,
    imgsz=640
)
```
